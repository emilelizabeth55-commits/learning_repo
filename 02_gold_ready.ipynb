```python
dim_customers = spark.table("emil_cat.silver.customers_clean")
dim_products = spark.table("emil_cat.silver.products_clean")

dim_customers.write.mode("overwrite").saveAsTable("emil_cat.gold.dim_customers")
dim_products.write.mode("overwrite").saveAsTable("emil_cat.gold.dim_products")
```

```python
from pyspark.sql import functions as F

orders_clean = spark.table("emil_cat.silver.orders_clean")
dim_products = spark.table("emil_cat.gold.dim_products")

fact_orders = (
    orders_clean.join(dim_products, on="product_code", how="inner")
    .withColumn("order_amount", F.col("quantity") * F.col("unit_price"))
    .select("order_id", "customer_id", "product_code", "quantity", "unit_price", "order_amount", "order_date")
)

fact_orders.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("emil_cat.gold.fact_orders")
display(fact_orders)

```
